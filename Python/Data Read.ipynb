### **3. Logical Validation**

```python
campaigns['start_date'] = pd.to_datetime(
    campaigns['start_date'],
    format='%d-%m-%Y'
)

campaigns['end_date'] = pd.to_datetime(
    campaigns['end_date'],
    format='%d-%m-%Y'
)

invalid_dates = campaigns[
    campaigns['start_date'] > campaigns['end_date']
]

print(f"Invalid campaign date records: {len(invalid_dates)}")
```

### **4. Primary Key Validation**

```python
print(f"NULL values: {campaigns['campaign_id'].isna().sum()}\n")
print(f"Duplicate valus: {campaigns['campaign_id'].duplicated().sum()}")
```

### **5. Foreign Key Validation**

```python
unreferenced_campaigns = campaigns[~campaigns['campaign_id'].isin(events['campaign_id'])]
print(f"unreferenced campaign: {len(unreferenced_campaigns)}")
```

### **6. Whitespace Validation**

```python
string_cols = campaigns.select_dtypes(include=['object', 'str']).columns

for col in string_cols:
    whitespace_rows = campaigns[
        campaigns[col] != campaigns[col].str.strip()]

    print(f"{col}: {len(whitespace_rows)} whitespace records")
```
